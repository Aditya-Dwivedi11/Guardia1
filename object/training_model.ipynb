```python
# =========================================================
# YOLOv8 FULL COCO (80 CLASSES) TRAINING + AUTO CLEANUP
# =========================================================

# 1️⃣ Install dependencies
!pip install -q ultralytics torch torchvision opencv-python

# 2️⃣ Imports
import os
import shutil
from ultralytics import YOLO

# 3️⃣ Dataset directory (Colab safe)
DATASET_DIR = "/content/yolo_datasets"
os.environ["ULTRALYTICS_DATASETS_DIR"] = DATASET_DIR

print("📁 Dataset directory:", DATASET_DIR)
```

```python
# 4️⃣ Load YOLOv8 model (best balance for surveillance)
model = YOLO("yolov8m.pt")  # m = accuracy + speed
model.train(
    data="coco.yaml",     # FULL COCO (not coco128)
    epochs=30,            # safe for Colab
    imgsz=640,
    batch=32,
    device="cuda",
    workers=2,

    optimizer="AdamW",
    lr0=0.001,
    weight_decay=0.0005,

    warmup_epochs=3,
    patience=10,
    close_mosaic=10,

    project="runs/surveillance",
    name="coco80_surveillance",
    pretrained=True,
    verbose=True
)

# 6️⃣ Validate model
metrics = model.val(
    data="coco.yaml",
    imgsz=640,
    batch=4,
    device="cuda"
)

print("\n========== MODEL METRICS ==========")
print(f"mAP@50     : {metrics.box.map50:.4f}")
print(f"mAP@50-95  : {metrics.box.map:.4f}")
print(f"Precision  : {metrics.box.mp:.4f}")
print(f"Recall     : {metrics.box.mr:.4f}")
print("==================================")

# # 7️⃣ DELETE DATASET TO FREE SPACE 🔥
# if os.path.exists(DATASET_DIR):
#     shutil.rmtree(DATASET_DIR)
#     print("\n🧹 Dataset deleted to free disk space!")

# 8️⃣ Final model path
print("\n✅ TRAINING COMPLETE")
print("📦 Best model saved at:")
print("runs/surveillance/coco80_surveillance/weights/best.pt")
```
