```python
import mediapipe as mp
import cv2
```

```python
mp_drawing = mp.solutions.drawing_utils
mp_holistic = mp.solutions.holistic
```

```python
cap = cv2.VideoCapture(0)

with mp_holistic.Holistic(min_detection_confidence=0.2,min_tracking_confidence=0.2) as fr:
    while cap.isOpened():
        ret,frame = cap.read()
        #Recolor
        img = cv2.cvtColor(frame,cv2.COLOR_BGR2RGB)
        #making detection
        result = fr.process(img)
        print(result.face_landmarks)
        #recolor image to rerender in screen with landmarks
        img = cv2.cvtColor(img,cv2.COLOR_RGB2BGR)
        #draw landmark in face
        # mp_drawing.draw_landmarks(img,result.face_landmarks,mp_holistic.FACEMESH_TESSELATION)

        #right hand
        mp_drawing.draw_landmarks(img,result.hands_connections,mp.solutions.hands)
        #left hand

        #pose 
        cv2.imshow('Raw webcam feed',img)
        if cv2.waitKey(10) & 0xFF==ord('q'):
            break
# cv2.waitKey(0)
cap.release()
cv2.destroyAllWindows()
print("Camera released safely.")
```
